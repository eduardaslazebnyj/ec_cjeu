# Historical European Commission antitrust decisions, 1964–2000

## Purpose

This notebook retrieves and parses the historical European Commission DG Competition
antitrust decision pages for 1964–2000 and downloads linked decision documents.

The notebook is deliberately kept compact:

1. cache the annual source pages;
2. parse the historical entries;
3. build and run document downloads;
4. save download progress to JSONL so interrupted runs can resume;
5. export concise CSV and Excel views.

The historical parsing logic is source-specific and therefore retained in detail.
Operational diagnostics stay under `logs/`; they are not publication data products.


```python
from __future__ import annotations

import hashlib
import json
import logging
import os
import re
import time
from collections import Counter, defaultdict
from datetime import datetime, timezone
from pathlib import Path
from urllib.parse import parse_qs, urldefrag, urljoin, urlparse

import pandas as pd
import requests
from bs4 import BeautifulSoup
from tqdm.auto import tqdm
```

## Repository paths and configuration

Paths written to tables are repository-relative. Set `DOWNLOAD_DOCUMENTS=False`
when you only want to rebuild the manifests from files already present locally.
Failed downloads are retried by default; set `RETRY_FAILED_DOWNLOADS=False`
to preserve their current state without retrying them.


```python
def find_project_root(start: Path | None = None) -> Path:
    override = os.environ.get("ECCJEU_PROJECT_ROOT", "").strip()
    if override:
        root = Path(override).expanduser().resolve()
        if not root.exists():
            raise FileNotFoundError(f"ECCJEU_PROJECT_ROOT does not exist: {root}")
        return root

    start = (start or Path.cwd()).resolve()
    candidates = [start, *start.parents]

    for candidate in candidates:
        if (candidate / ".git").exists():
            return candidate

    expected = {"code", "data", "output", "logs"}
    for candidate in candidates:
        present = {name for name in expected if (candidate / name).exists()}
        if {"code", "data"}.issubset(present) and len(present) >= 3:
            return candidate

    raise RuntimeError(
        "Could not identify the repository root. Start Jupyter from within the repository "
        "or set ECCJEU_PROJECT_ROOT to the repository directory."
    )


PROJECT_ROOT = find_project_root()

YEAR_START = 1964
YEAR_END = 2000
YEARS = list(range(YEAR_START, YEAR_END + 1))
BASE_URL = "http://ec.europa.eu/competition/antitrust/closed/en/{year}.html"

FORCE_REFETCH_YEAR_PAGES = False
DOWNLOAD_DOCUMENTS = True
OVERWRITE_EXISTING_FILES = False
RETRY_FAILED_DOWNLOADS = True

SOURCE_REQUEST_TIMEOUT_SECONDS = 30
SOURCE_REQUEST_SLEEP_SECONDS = 0.25
DOCUMENT_REQUEST_TIMEOUT_SECONDS = 45
DOCUMENT_REQUEST_SLEEP_SECONDS = 0.30
MIN_SOURCE_PAGE_BYTES = 200

RAW_DIR = PROJECT_ROOT / "data" / "raw" / "com_legacy"
YEAR_HTML_DIR = RAW_DIR / "html_year_pages"
FILES_DIR = RAW_DIR / "files"
HTML_FILES_DIR = FILES_DIR / "html"
PDF_FILES_DIR = FILES_DIR / "pdfs"

OUTPUT_DIR = PROJECT_ROOT / "output" / "com_legacy"
LOG_DIR = PROJECT_ROOT / "logs" / "com_legacy"

SOURCE_PAGE_METADATA_PATH = RAW_DIR / "source_pages_metadata.jsonl"
DOWNLOAD_STATE_JSONL = LOG_DIR / "download_state.jsonl"

RECORDS_CSV = OUTPUT_DIR / "com_legacy_records.csv"
DOCUMENTS_CSV = OUTPUT_DIR / "com_legacy_documents.csv"
WORKBOOK_XLSX = OUTPUT_DIR / "com_legacy.xlsx"

for directory in [YEAR_HTML_DIR, HTML_FILES_DIR, PDF_FILES_DIR, OUTPUT_DIR, LOG_DIR]:
    directory.mkdir(parents=True, exist_ok=True)


def repo_relative(path: Path) -> str:
    return path.resolve().relative_to(PROJECT_ROOT.resolve()).as_posix()


log_path = LOG_DIR / f"01_download_com_legacy_{datetime.now(timezone.utc):%Y%m%dT%H%M%SZ}.log"
logger = logging.getLogger("01_download_com_legacy")
logger.handlers.clear()
logger.setLevel(logging.INFO)
_handler = logging.FileHandler(log_path, encoding="utf-8")
_handler.setFormatter(logging.Formatter("%(asctime)s | %(levelname)s | %(message)s"))
logger.addHandler(_handler)
logger.propagate = False

print("Project root: .")
print(f"Raw source pages: {repo_relative(YEAR_HTML_DIR)}")
print(f"Publication output: {repo_relative(OUTPUT_DIR)}")
print(f"Download progress: {repo_relative(DOWNLOAD_STATE_JSONL)}")
print(f"Document retrieval enabled: {DOWNLOAD_DOCUMENTS}")
```

## 1. Retrieve and cache annual Commission pages

The 37 annual pages are cached under `data/raw/com_legacy/html_year_pages/`.
Existing pages are reused unless `FORCE_REFETCH_YEAR_PAGES=True`.

A small source-page metadata JSONL is retained with the raw cache so that retrieval
metadata are not invented for files that pre-date the current run.


```python
def utc_now_iso() -> str:
    return datetime.now(timezone.utc).replace(microsecond=0).isoformat().replace("+00:00", "Z")


def sha256_bytes(data: bytes) -> str:
    return hashlib.sha256(data).hexdigest()


def sha256_file(path: Path) -> str:
    h = hashlib.sha256()
    with path.open("rb") as fh:
        for chunk in iter(lambda: fh.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()


def year_cache_path(year: int) -> Path:
    return YEAR_HTML_DIR / f"ec_competition_antitrust_closed_{year}.html"


def read_jsonl_index(path: Path, key: str) -> dict:
    out = {}
    if not path.exists():
        return out
    for raw in path.read_text(encoding="utf-8", errors="replace").splitlines():
        raw = raw.strip()
        if not raw:
            continue
        try:
            obj = json.loads(raw)
        except json.JSONDecodeError:
            continue
        if key in obj:
            out[str(obj[key])] = obj
    return out


def write_jsonl_rows(path: Path, rows: list[dict]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8") as fh:
        for row in rows:
            fh.write(json.dumps(row, ensure_ascii=False, sort_keys=True) + "\n")


SOURCE_METADATA_INDEX = read_jsonl_index(SOURCE_PAGE_METADATA_PATH, "source_year")

SESSION = requests.Session()
SESSION.headers.update({
    "User-Agent": "Mozilla/5.0 (compatible; ZHAW-EU-Antitrust-Research/1.0)",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.9,de;q=0.7",
})


def cached_page_provenance(year: int, path: Path, source_url: str) -> dict:
    data = path.read_bytes()
    previous = SOURCE_METADATA_INDEX.get(str(year), {})
    previous_hash = previous.get("sha256", "")
    same_snapshot = bool(previous_hash) and previous_hash == sha256_bytes(data)
    return {
        "source_year": year,
        "source_url": source_url,
        "retrieved_at_utc": previous.get("retrieved_at_utc", "") if same_snapshot else "",
        "http_status": previous.get("http_status", pd.NA) if same_snapshot else pd.NA,
        "final_url": previous.get("final_url", source_url) if same_snapshot else source_url,
        "relative_cache_path": repo_relative(path),
        "file_size_bytes": len(data),
        "sha256": sha256_bytes(data),
        "acquisition_status": "cached_with_metadata" if same_snapshot else "cached_existing_without_metadata",
        "error": "",
    }


def retrieve_year_page(year: int, force: bool = False) -> dict:
    source_url = BASE_URL.format(year=year)
    path = year_cache_path(year)

    if path.exists() and not force:
        return cached_page_provenance(year, path, source_url)

    try:
        response = SESSION.get(source_url, timeout=SOURCE_REQUEST_TIMEOUT_SECONDS, allow_redirects=True)
        content = response.content or b""
        retrieved_at = utc_now_iso()
        if response.status_code == 200 and len(content) >= MIN_SOURCE_PAGE_BYTES:
            path.write_bytes(content)
            result = {
                "source_year": year,
                "source_url": source_url,
                "retrieved_at_utc": retrieved_at,
                "http_status": response.status_code,
                "final_url": response.url,
                "relative_cache_path": repo_relative(path),
                "file_size_bytes": len(content),
                "sha256": sha256_bytes(content),
                "acquisition_status": "retrieved",
                "error": "",
            }
        else:
            result = {
                "source_year": year,
                "source_url": source_url,
                "retrieved_at_utc": retrieved_at,
                "http_status": response.status_code,
                "final_url": response.url,
                "relative_cache_path": "",
                "file_size_bytes": len(content),
                "sha256": "",
                "acquisition_status": "unavailable",
                "error": f"HTTP {response.status_code}; bytes={len(content)}",
            }
    except Exception as exc:
        logger.exception("Failed to retrieve annual source page for %s", year)
        result = {
            "source_year": year,
            "source_url": source_url,
            "retrieved_at_utc": utc_now_iso(),
            "http_status": pd.NA,
            "final_url": source_url,
            "relative_cache_path": "",
            "file_size_bytes": 0,
            "sha256": "",
            "acquisition_status": "unavailable",
            "error": repr(exc),
        }

    time.sleep(SOURCE_REQUEST_SLEEP_SECONDS)
    return result


source_page_rows = [retrieve_year_page(year, force=FORCE_REFETCH_YEAR_PAGES) for year in YEARS]
source_pages_df = pd.DataFrame(source_page_rows)

# Persist only provenance generated by this notebook; do not fabricate metadata for old cache snapshots.
metadata_rows = []
for row in source_page_rows:
    if row["acquisition_status"] in {"retrieved", "cached_with_metadata"}:
        metadata_rows.append({k: row[k] for k in [
            "source_year", "source_url", "retrieved_at_utc", "http_status",
            "final_url", "relative_cache_path", "file_size_bytes", "sha256"
        ]})
if metadata_rows:
    merged = {str(k): v for k, v in SOURCE_METADATA_INDEX.items()}
    for row in metadata_rows:
        merged[str(row["source_year"])] = row
    write_jsonl_rows(SOURCE_PAGE_METADATA_PATH, [merged[k] for k in sorted(merged, key=int)])

available_pages = source_pages_df["relative_cache_path"].astype(str).ne("").sum()
print(f"Annual source pages available in cache: {available_pages}/{len(YEARS)}")
```

## 2. Parse historical Commission entries

The historical pages are not uniform. The parser therefore keeps the source-specific
logic from the original notebook: line-oriented parsing first, DOM-block fallback
where necessary, CELEX and case-number extraction, Official Journal parsing,
and exclusion of obvious navigation/footer links.

This is the part of the pipeline where complexity is useful because the source itself
is irregular.


```python
DATE_RE = re.compile(r"\b\d{1,2}[./]\d{1,2}[./]\d{4}\b")
CASE_NO_RE = re.compile(r"\b([A-Z]{1,5})\s*/\s*([0-9]{1,7}[A-Z]?)\b")
CELEX_GENERIC_RE = re.compile(r"\b3\s*\d{2,4}\s*[A-Z]\s*\d{2}\s*\d{2}\b", re.I)
CELEX_COMPLETE_RE = re.compile(r"^3\d{4}[A-Z]\d{4}$", re.I)

DECISION_WORDING_RE = re.compile(
    r"\b(infring(?:e)?ment|prohibition|exemption|negative\s+clearance|non-opposition|"
    r"interim\s+measures?|rejection\s+of\s+complaint|decision|art\.\s*1[1-9])\b",
    re.I,
)
DETAIL_MARKERS = (
    "official journal", "celex no", "press release", "full text",
    "text in pdf format", "text in pdf", "text pdf",
)


def clean_text(value: object) -> str:
    if value is None:
        return ""
    return re.sub(r"\s+", " ", str(value).replace("\xa0", " ")).strip()


def dedupe_keep_order(values: list[str]) -> list[str]:
    out, seen = [], set()
    for value in values:
        value = clean_text(value)
        if value and value not in seen:
            seen.add(value)
            out.append(value)
    return out


def decode_html_bytes(data: bytes) -> str:
    try:
        return data.decode("utf-8")
    except UnicodeDecodeError:
        return data.decode("latin-1", errors="replace")


def parse_source_date(raw: str) -> str:
    raw = clean_text(raw)
    if not raw:
        return ""
    for fmt in ("%d.%m.%Y", "%d/%m/%Y"):
        try:
            return datetime.strptime(raw, fmt).date().isoformat()
        except ValueError:
            continue
    return ""


def normalize_celex(value: str | None) -> str:
    return re.sub(r"[^0-9A-Za-z]", "", value or "").upper()


def complete_celex(value: str | None) -> str:
    celex = normalize_celex(value)
    if not celex:
        return ""
    if CELEX_COMPLETE_RE.fullmatch(celex):
        return celex
    match = re.fullmatch(r"3(\d{2})([A-Z])(\d{4})", celex)
    if match:
        yy, letter, number = match.groups()
        year = f"19{yy}" if int(yy) >= 50 else f"20{yy}"
        return f"3{year}{letter}{number}"
    return celex


def canonical_url_without_fragment(url: str) -> str:
    base, _ = urldefrag(url)
    parsed = urlparse(base)
    scheme = parsed.scheme.lower()
    netloc = parsed.netloc.lower()
    path = re.sub(r"/+", "/", parsed.path or "/").rstrip("/") or "/"
    query = parsed.query
    return f"{scheme}://{netloc}{path}" + (f"?{query}" if query else "")


def is_navigation_link(link: dict, source_url: str) -> bool:
    href = clean_text(link.get("href", ""))
    url = clean_text(link.get("url", ""))
    label = clean_text(link.get("text", "")).lower()
    href_low = href.lower()

    if not href or href_low.startswith(("javascript:", "mailto:")):
        return True
    if href_low in {"#top", "#previous", "#next"}:
        return True
    if label in {"top", "previous", "next", "search", "home", "contact", "sitemap"}:
        return True

    resolved = url or urljoin(source_url, href)
    if canonical_url_without_fragment(resolved) == canonical_url_without_fragment(source_url):
        return True

    path = urlparse(resolved).path.lower()
    if path.endswith("/competition/search.html") or "/competition/search" in path:
        return True
    chrome_terms = ("/contact", "/sitemap", "/legal", "/cookies", "/privacy", "/index.html")
    if any(term in path for term in chrome_terms):
        return True
    if re.search(r"/(?:previous|next)(?:\.|/|$)", path):
        return True
    if path.endswith((".gif", ".jpg", ".jpeg", ".png", ".svg")):
        return True
    return False


def extract_links(fragment_html: str, source_url: str) -> list[dict]:
    soup = BeautifulSoup(fragment_html, "html.parser")
    links = []
    for anchor in soup.find_all("a", href=True):
        href = anchor.get("href", "")
        link = {
            "text": clean_text(anchor.get_text(" ", strip=True)),
            "href": href,
            "url": urljoin(source_url, href),
        }
        link["is_navigation"] = is_navigation_link(link, source_url)
        links.append(link)
    # De-duplicate without losing source order.
    unique, seen = [], set()
    for link in links:
        key = (link["text"], link["href"], link["url"])
        if key not in seen:
            seen.add(key)
            unique.append(link)
    return unique


def filtered_source_links(links: list[dict]) -> list[dict]:
    return [{k: v for k, v in link.items() if k != "is_navigation"} for link in links if not link.get("is_navigation")]


def split_html_into_logical_lines(html: str, source_url: str) -> list[dict]:
    cleaned = re.sub(r"<script\b.*?</script>", "", html, flags=re.I | re.S)
    cleaned = re.sub(r"<style\b.*?</style>", "", cleaned, flags=re.I | re.S)
    # Add synthetic line breaks after block elements while preserving each fragment's own links.
    cleaned = re.sub(r"</(p|tr|li|h[1-6])\s*>", lambda m: m.group(0) + "<br/>", cleaned, flags=re.I)
    parts = re.split(r"<br\s*/?>", cleaned, flags=re.I)

    rows = []
    for index, fragment in enumerate(parts):
        soup = BeautifulSoup(fragment, "html.parser")
        text = clean_text(soup.get_text(" ", strip=True))
        if not text:
            continue
        all_links = extract_links(fragment, source_url)
        rows.append({
            "source_line_index": index,
            "source_line_text": text,
            "source_line_html": fragment.strip(),
            "links": filtered_source_links(all_links),
            "navigation_links": [
                {k: v for k, v in link.items() if k != "is_navigation"}
                for link in all_links if link.get("is_navigation")
            ],
            "segmentation": "line",
        })
    return rows


def dom_block_fallback(html: str, source_url: str) -> list[dict]:
    soup = BeautifulSoup(html, "html.parser")
    for tag in soup(["script", "style"]):
        tag.decompose()

    blocks = []
    tags = soup.find_all(["tr", "p", "li"])
    if not tags:
        tags = soup.find_all("div")

    for index, tag in enumerate(tags):
        text = clean_text(tag.get_text(" ", strip=True))
        if not text:
            continue
        all_links = extract_links(str(tag), source_url)
        blocks.append({
            "source_line_index": index,
            "source_line_text": text,
            "source_line_html": str(tag),
            "links": filtered_source_links(all_links),
            "navigation_links": [
                {k: v for k, v in link.items() if k != "is_navigation"}
                for link in all_links if link.get("is_navigation")
            ],
            "segmentation": "dom_block_fallback",
        })
    return blocks


def is_navigation_or_header_line(text: str) -> bool:
    low = clean_text(text).lower()
    if not low:
        return True
    exact = {
        "previous", "next", "top", "search", "home", "contact", "sitemap",
        "legal notice", "copyright notice", "important legal notice",
    }
    if low in exact:
        return True
    prefixes = (
        "index of antitrust formal decisions",
        "decisions art.",
        "last page update",
        "the information on this site",
    )
    return low.startswith(prefixes)


def is_detail_line(text: str) -> bool:
    low = clean_text(text).lower()
    if low in {"text", "pdf", "html", "text pdf"}:
        return True
    return any(marker in low for marker in DETAIL_MARKERS)


def looks_like_case_line(text: str) -> bool:
    text = clean_text(text)
    if is_navigation_or_header_line(text) or is_detail_line(text):
        return False
    if DATE_RE.search(text) or CASE_NO_RE.search(text) or CELEX_GENERIC_RE.search(text):
        return True
    return bool(len(text) >= 8 and DECISION_WORDING_RE.search(text))


def extract_celex_from_text(text: str) -> tuple[list[str], list[str]]:
    raw_values = []
    text = clean_text(text)
    for match in re.finditer(r"Celex\s+No\.\s*:\s*([0-9A-Za-z\s]+)", text, flags=re.I):
        candidate = re.split(r"\s+-\s+|\s+[A-Z]{1,5}\s*/", match.group(1), maxsplit=1, flags=re.I)[0]
        value = normalize_celex(candidate)
        if re.fullmatch(r"3\d{2,4}[A-Z]\d{4}", value):
            raw_values.append(value)
    for match in CELEX_GENERIC_RE.finditer(text):
        value = normalize_celex(match.group(0))
        if re.fullmatch(r"3\d{2,4}[A-Z]\d{4}", value):
            raw_values.append(value)
    raw_values = dedupe_keep_order(raw_values)
    complete = dedupe_keep_order([complete_celex(value) for value in raw_values])
    return raw_values, complete


def extract_celex_from_links(links: list[dict]) -> tuple[list[str], list[str], list[str]]:
    raw_values, urls = [], []
    for link in links:
        href = link.get("href", "") or ""
        url = link.get("url", "") or ""
        text = link.get("text", "") or ""
        haystack = f"{href} {url} {text}"
        if "celexnumdoc" not in haystack.lower() and "numdoc=" not in haystack.lower() and "celex:" not in haystack.lower():
            continue
        if url:
            urls.append(url)
        query = parse_qs(urlparse(href).query)
        for value in (query.get("numdoc") or query.get("uri") or []):
            candidate = normalize_celex(value.replace("CELEX:", ""))
            if candidate:
                raw_values.append(candidate)
        for match in re.finditer(r"(?:numdoc=|CELEX:)([0-9A-Za-z]+)", href, flags=re.I):
            raw_values.append(normalize_celex(match.group(1)))
        visible = normalize_celex(text)
        if re.fullmatch(r"3\d{2,4}[A-Z]\d{4}", visible):
            raw_values.append(visible)
    raw_values = dedupe_keep_order(raw_values)
    complete = dedupe_keep_order([complete_celex(value) for value in raw_values])
    return raw_values, complete, dedupe_keep_order(urls)


def extract_case_numbers(text: str) -> list[str]:
    return dedupe_keep_order([f"{prefix.upper()}/{number}" for prefix, number in CASE_NO_RE.findall(text or "")])


def extract_official_journal(text: str) -> dict:
    result = {
        "official_journal_raw": "",
        "oj_series": "",
        "oj_issue": "",
        "oj_date_raw": "",
        "oj_date": "",
        "oj_page": "",
    }
    match = re.search(
        r"Official Journal\s*:\s*(.*?)(?=\s+Celex\s+No\.|\s+-\s+\b[A-Z]{1,5}\s*/\s*\d|$)",
        text or "",
        flags=re.I,
    )
    if not match:
        return result
    raw = clean_text(match.group(1)).strip(" -")
    result["official_journal_raw"] = raw
    series = re.match(r"\s*([LC])\b", raw, flags=re.I)
    if series:
        result["oj_series"] = series.group(1).upper()
    issue = re.match(r"\s*[LC]\s*(\d+)\s*-", raw, flags=re.I)
    if issue:
        result["oj_issue"] = issue.group(1)
    date_match = DATE_RE.search(raw)
    if date_match:
        result["oj_date_raw"] = date_match.group(0)
        result["oj_date"] = parse_source_date(date_match.group(0))
    page_match = re.search(r"Page\s*:\s*([0-9]+(?:\s*[-–]\s*[0-9]+)?)", raw, flags=re.I)
    if page_match:
        result["oj_page"] = clean_text(page_match.group(1))
    return result


def strip_metadata_from_decision_type(text: str) -> str:
    positions = []
    for pattern in [
        r"Official Journal\s*:", r"Celex\s+No\.\s*:",
        r"\b3\s*\d{2,4}\s*[A-Z]\s*\d{2}\s*\d{2}\b",
        r"\s+-\s+\b[A-Z]{1,5}\s*/\s*\d{1,7}",
        r"\b[A-Z]{1,5}\s*/\s*\d{1,7}", r"\s+Press Release\b",
    ]:
        match = re.search(pattern, text, flags=re.I)
        if match:
            positions.append(match.start())
    end = min(positions) if positions else len(text)
    return clean_text(text[:end]).strip(" -")


def parse_title_date_type(text: str) -> dict:
    text = clean_text(text)
    date_match = DATE_RE.search(text)
    if date_match:
        raw = date_match.group(0)
        return {
            "title": clean_text(text[:date_match.start()]),
            "decision_date_raw": raw,
            "decision_date": parse_source_date(raw),
            "decision_type": strip_metadata_from_decision_type(text[date_match.end():]),
        }

    positions = []
    for pattern in [CELEX_GENERIC_RE.pattern, CASE_NO_RE.pattern, r"Official Journal\s*:", r"Celex\s+No\.\s*:"]:
        match = re.search(pattern, text, flags=re.I)
        if match:
            positions.append(match.start())
    end = min(positions) if positions else len(text)
    return {
        "title": clean_text(text[:end]).strip(" -") or text,
        "decision_date_raw": "",
        "decision_date": "",
        "decision_type": "",
    }


def classify_links(links: list[dict], line_context: str = "") -> list[dict]:
    classified = []
    for link in links:
        url = clean_text(link.get("url", ""))
        href = clean_text(link.get("href", ""))
        label = clean_text(link.get("text", ""))
        haystack = f"{label} {line_context} {href} {url}".lower()
        path = urlparse(url).path.lower()

        if "celexnumdoc" in haystack or "numdoc=" in haystack or "celex:" in haystack:
            category = "celex"
        elif "press release" in haystack or "rapid" in haystack or "p_action.gettxt" in haystack:
            category = "press_release"
        elif "official journal" in haystack or "uriserv:oj" in haystack or "toc=oj" in haystack:
            category = "official_journal"
        elif "full text" in haystack or "text in pdf" in haystack:
            category = "full_text"
        elif path.endswith(".pdf") and not path.endswith(("search.pdf",)):
            category = "full_text"
        else:
            category = "other"
        classified.append({**link, "classification": category})
    return classified


def record_hash_payload(year: int, source_url: str, source_text: str, detail_texts: list[str], links: list[dict]) -> str:
    payload = {
        "source_year": int(year),
        "source_url": canonical_url_without_fragment(source_url),
        "source_text": clean_text(source_text),
        "detail_texts": [clean_text(x) for x in detail_texts],
        "links": [
            {"text": clean_text(x.get("text", "")), "url": clean_text(x.get("url", "")), "classification": x.get("classification", "")}
            for x in links
        ],
    }
    encoded = json.dumps(payload, ensure_ascii=False, sort_keys=True, separators=(",", ":")).encode("utf-8")
    return hashlib.sha256(encoded).hexdigest()


def parse_case_record(year: int, source_url: str, case_line: dict, detail_lines: list[dict]) -> dict | None:
    source_text = clean_text(case_line.get("source_line_text", ""))
    if not looks_like_case_line(source_text):
        return None

    detail_texts = [clean_text(line.get("source_line_text", "")) for line in detail_lines]
    combined_text = clean_text(" ".join([source_text, *detail_texts]))

    links = []
    for line in [case_line, *detail_lines]:
        context = clean_text(line.get("source_line_text", ""))
        links.extend(classify_links(line.get("links", []), context))
    # exact de-duplication
    unique_links, seen = [], set()
    for link in links:
        key = (link.get("text", ""), link.get("url", ""), link.get("classification", ""))
        if key not in seen:
            seen.add(key)
            unique_links.append(link)
    links = unique_links

    text_raw, text_complete = extract_celex_from_text(combined_text)
    link_raw, link_complete, celex_urls = extract_celex_from_links(links)
    celex_raw = dedupe_keep_order([*link_raw, *text_raw])
    celex_ids = dedupe_keep_order([*link_complete, *text_complete])
    case_numbers = extract_case_numbers(combined_text)
    oj = extract_official_journal(combined_text)
    parsed = parse_title_date_type(source_text)

    source_record_hash = record_hash_payload(year, source_url, source_text, detail_texts, links)
    return {
        "source_record_hash": source_record_hash,
        "source": {
            "year": year,
            "url": source_url,
            "line_index": case_line.get("source_line_index"),
            "segmentation": case_line.get("segmentation", ""),
            "text": source_text,
            "html": case_line.get("source_line_html", ""),
        },
        "detail_lines": [
            {
                "line_index": line.get("source_line_index"),
                "segmentation": line.get("segmentation", ""),
                "text": clean_text(line.get("source_line_text", "")),
                "html": line.get("source_line_html", ""),
            }
            for line in detail_lines
        ],
        "parsed": {
            **parsed,
            "official_journal": oj,
            "case_numbers": case_numbers,
            "celex_raw": celex_raw,
            "celex": celex_ids,
        },
        "links": links,
        "celex_source_urls": celex_urls,
    }


def parse_line_sequence(year: int, source_url: str, lines: list[dict]) -> tuple[list[dict], list[dict]]:
    records, ignored = [], []
    current_case, current_details = None, []

    def flush_current():
        nonlocal current_case, current_details
        if current_case is not None:
            record = parse_case_record(year, source_url, current_case, current_details)
            if record is not None:
                records.append(record)
        current_case, current_details = None, []

    for line in lines:
        text = clean_text(line.get("source_line_text", ""))
        if is_navigation_or_header_line(text):
            flush_current()
            continue
        if is_detail_line(text):
            if current_case is not None:
                current_details.append(line)
            else:
                ignored.append({"year": year, "line_index": line.get("source_line_index"), "reason": "orphan_detail", "text": text})
            continue
        if looks_like_case_line(text):
            flush_current()
            current_case = line
            current_details = []
            continue
        ignored.append({"year": year, "line_index": line.get("source_line_index"), "reason": "unclassified", "text": text})

    flush_current()
    return records, ignored


def page_diagnostics(html: str) -> dict:
    soup = BeautifulSoup(html, "html.parser")
    text = clean_text(soup.get_text(" ", strip=True))
    title = clean_text(soup.title.get_text(" ", strip=True)) if soup.title else ""
    return {
        "html_title": title,
        "text_characters": len(text),
        "date_hits": len(DATE_RE.findall(text)),
        "case_number_hits": len(CASE_NO_RE.findall(text)),
        "celex_hits": len(CELEX_GENERIC_RE.findall(text)),
        "official_journal_hits": text.lower().count("official journal"),
        "legacy_index_marker": "index of antitrust formal decisions" in text.lower(),
    }


def parse_year_page(year: int, source_url: str, html: str) -> tuple[list[dict], list[dict], dict]:
    diagnostics = page_diagnostics(html)
    primary_lines = split_html_into_logical_lines(html, source_url)
    records, ignored = parse_line_sequence(year, source_url, primary_lines)
    strategy = "line"

    if not records:
        fallback_lines = dom_block_fallback(html, source_url)
        records, fallback_ignored = parse_line_sequence(year, source_url, fallback_lines)
        ignored.extend(fallback_ignored)
        strategy = "dom_block_fallback" if records else "no_records"

    diagnostics["parse_strategy"] = strategy
    diagnostics["logical_line_count"] = len(primary_lines)
    diagnostics["record_count"] = len(records)
    return records, ignored, diagnostics
```

## 3. Construct the source-record table

Rich nested parser objects remain in memory while this notebook runs. The exported
record table is intentionally flatter and easier to review.

The notebook reports any year that contains decision-like source signals but yields
no parsed records; those years deserve manual parser review.


```python
parsed_records: list[dict] = []
ignored_lines: list[dict] = []
parse_status_by_year: dict[int, dict] = {}

for row in source_page_rows:
    year = int(row["source_year"])
    cache_rel = clean_text(row.get("relative_cache_path", ""))

    if not cache_rel:
        parse_status_by_year[year] = {
            "parse_status": "source_unavailable",
            "record_count": 0,
            "parse_notes": clean_text(row.get("error", "")),
        }
        continue

    cache_path = PROJECT_ROOT / cache_rel
    html = decode_html_bytes(cache_path.read_bytes())
    year_records, year_ignored, diagnostics = parse_year_page(year, row["source_url"], html)

    parsed_records.extend(year_records)
    ignored_lines.extend(year_ignored)

    if year_records:
        status = "parsed"
        notes = f"strategy={diagnostics['parse_strategy']}"
    else:
        positive_signals = (
            diagnostics["date_hits"]
            + diagnostics["case_number_hits"]
            + diagnostics["celex_hits"]
        )
        status = "parser_review_required" if positive_signals else "no_decision_structure_detected"
        notes = json.dumps(diagnostics, ensure_ascii=False, sort_keys=True)

    parse_status_by_year[year] = {
        "parse_status": status,
        "record_count": len(year_records),
        "parse_notes": notes,
    }

# Stable IDs are retained internally, but the full record hash is not exported.
hash_totals = Counter(record["source_record_hash"] for record in parsed_records)
hash_seen = Counter()

for record in parsed_records:
    digest = record["source_record_hash"]
    year = int(record["source"]["year"])
    hash_seen[digest] += 1
    base = f"COMLEG_{year}_{digest[:16]}"
    record["source_entry_id"] = (
        base if hash_totals[digest] == 1 else f"{base}_{hash_seen[digest]:02d}"
    )

source_pages_df["parse_status"] = source_pages_df["source_year"].map(
    lambda y: parse_status_by_year[int(y)]["parse_status"]
)
source_pages_df["record_count"] = source_pages_df["source_year"].map(
    lambda y: parse_status_by_year[int(y)]["record_count"]
)
source_pages_df["parse_notes"] = source_pages_df["source_year"].map(
    lambda y: parse_status_by_year[int(y)]["parse_notes"]
)

if ignored_lines:
    pd.DataFrame(ignored_lines).to_csv(
        LOG_DIR / "parse_ignored_lines.csv",
        index=False,
    )

record_rows = []
for record in parsed_records:
    parsed = record["parsed"]
    record_rows.append({
        "source_entry_id": record["source_entry_id"],
        "source_year": record["source"]["year"],
        "source_url": record["source"]["url"],
        "title": parsed["title"],
        "decision_date_raw": parsed["decision_date_raw"],
        "decision_date": parsed["decision_date"],
        "decision_type": parsed["decision_type"],
        "official_journal_raw": parsed["official_journal"]["official_journal_raw"],
        "case_numbers": "; ".join(parsed["case_numbers"]),
        "celex_ids": "; ".join(parsed["celex"]),
        "document_count": 0,
    })

records_columns = [
    "source_entry_id",
    "source_year",
    "source_url",
    "title",
    "decision_date_raw",
    "decision_date",
    "decision_type",
    "official_journal_raw",
    "case_numbers",
    "celex_ids",
    "document_count",
]
records_df = pd.DataFrame(record_rows, columns=records_columns)

if not records_df.empty:
    sort_key = pd.to_datetime(records_df["decision_date"], errors="coerce")
    records_df = (
        records_df.assign(_sort_date=sort_key)
        .sort_values(["source_year", "_sort_date", "source_entry_id"], na_position="last")
        .drop(columns="_sort_date")
        .reset_index(drop=True)
    )

available_years = int(source_pages_df["relative_cache_path"].astype(str).ne("").sum())
parsed_years = int((source_pages_df["parse_status"] == "parsed").sum())
review_years = source_pages_df.loc[
    source_pages_df["parse_status"] == "parser_review_required", "source_year"
].astype(int).tolist()

print(f"Annual source pages available: {available_years}/{len(YEARS)}")
print(f"Annual source pages parsed: {parsed_years}/{len(YEARS)}")
print(f"Parsed source records: {len(records_df):,}")
print(f"Years requiring parser review: {review_years if review_years else 'none'}")
```

## 4. Download documents with resumable progress

`logs/com_legacy/download_state.jsonl` is the persistent operational checkpoint.

One row is appended when an attempt is completed. On a later run, successful attempts
whose files still exist are skipped. Failed attempts can be retried. This JSONL is for
resumability and troubleshooting, not for publication.


```python
def safe_filename(value: str) -> str:
    value = re.sub(r"[^A-Za-z0-9._-]+", "_", value or "").strip("._")
    return value or "document"


def eurlex_download_url(celex: str, language: str, file_format: str) -> str:
    fmt = "HTML" if file_format.lower() == "html" else "PDF"
    return (
        f"https://eur-lex.europa.eu/legal-content/"
        f"{language.upper()}/TXT/{fmt}/?uri=CELEX:{celex}"
    )


def infer_link_format(url: str) -> str:
    path = urlparse(url).path.lower()
    return "pdf" if ".pdf" in path else "html"


def infer_link_language(url: str) -> str:
    path = urlparse(url).path
    for pattern in [
        r"/([A-Za-z]{2})/[^/]*$",
        r"[_-]([A-Za-z]{2})(?:\.[A-Za-z0-9]+)?$",
    ]:
        match = re.search(pattern, path)
        if match:
            return match.group(1).upper()
    return "unknown"


def celex_output_path(celex: str, language: str, file_format: str) -> Path:
    folder = HTML_FILES_DIR if file_format == "html" else PDF_FILES_DIR
    return folder / f"{safe_filename(celex)}_{language.upper()}.{file_format}"


def direct_output_path(source_entry_id: str, url: str, file_format: str) -> Path:
    folder = HTML_FILES_DIR if file_format == "html" else PDF_FILES_DIR
    basename = Path(urlparse(url).path).name or f"document.{file_format}"
    return folder / f"{safe_filename(source_entry_id)}__{safe_filename(basename)}"


def valid_download(content: bytes, file_format: str, content_type: str = "") -> bool:
    if file_format == "pdf":
        return (
            len(content) > 1000
            and (
                content.startswith(b"%PDF")
                or "application/pdf" in content_type.lower()
            )
        )

    if len(content) <= 500:
        return False

    text = content[:20000].decode("utf-8", errors="ignore").lower()
    bad = (
        "request could not be satisfied",
        "not found",
        "verify that you're not a robot",
    )
    return not any(term in text for term in bad)


def attempt_id(attempt: dict) -> str:
    payload = "|".join([
        attempt.get("source_entry_id", ""),
        attempt.get("source_type", ""),
        attempt.get("celex", ""),
        attempt.get("language", ""),
        attempt.get("file_format", ""),
        attempt.get("source_url", ""),
    ])
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()[:20]


def build_download_attempts(records: list[dict]) -> list[dict]:
    attempts = []
    seen = set()

    for record in records:
        entry_id = record["source_entry_id"]

        for celex in record["parsed"]["celex"]:
            for language in ("EN", "DE"):
                for file_format in ("html", "pdf"):
                    url = eurlex_download_url(celex, language, file_format)
                    attempt = {
                        "source_entry_id": entry_id,
                        "celex": celex,
                        "source_type": "celex_eurlex",
                        "language": language,
                        "file_format": file_format,
                        "source_url": url,
                    }
                    attempt["attempt_id"] = attempt_id(attempt)
                    if attempt["attempt_id"] not in seen:
                        seen.add(attempt["attempt_id"])
                        attempts.append(attempt)

        for link in record["links"]:
            classification = link.get("classification")
            if classification not in {"full_text", "official_journal"}:
                continue

            url = link.get("url", "")
            file_format = infer_link_format(url)
            attempt = {
                "source_entry_id": entry_id,
                "celex": "",
                "source_type": (
                    "full_text" if classification == "full_text" else "official_journal"
                ),
                "language": infer_link_language(url),
                "file_format": file_format,
                "source_url": url,
            }
            attempt["attempt_id"] = attempt_id(attempt)
            if attempt["attempt_id"] not in seen:
                seen.add(attempt["attempt_id"])
                attempts.append(attempt)

    return attempts


def existing_exact_attempt_path(attempt: dict) -> Path | None:
    if attempt["source_type"] == "celex_eurlex":
        path = celex_output_path(
            attempt["celex"],
            attempt["language"],
            attempt["file_format"],
        )
        return path if path.exists() and path.stat().st_size > 0 else None

    folder = HTML_FILES_DIR if attempt["file_format"] == "html" else PDF_FILES_DIR
    basename = Path(urlparse(attempt["source_url"]).path).name

    # Backward compatibility with older direct-download filenames.
    if basename:
        old_path = folder / safe_filename(basename)
        if old_path.exists() and old_path.stat().st_size > 0:
            return old_path

    new_path = direct_output_path(
        attempt["source_entry_id"],
        attempt["source_url"],
        attempt["file_format"],
    )
    return new_path if new_path.exists() and new_path.stat().st_size > 0 else None


def append_jsonl_row(path: Path, row: dict) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("a", encoding="utf-8") as fh:
        fh.write(json.dumps(row, ensure_ascii=False, sort_keys=True) + "\n")
        fh.flush()


def load_download_state(path: Path) -> dict[str, dict]:
    return read_jsonl_index(path, "attempt_id")


def result_path_exists(result: dict) -> bool:
    rel = clean_text(result.get("relative_path", ""))
    return bool(rel) and (PROJECT_ROOT / rel).exists()


def run_document_downloads(
    attempts: list[dict],
) -> tuple[list[dict], dict[str, dict]]:
    """
    Process download attempts with a JSONL checkpoint.

    The JSONL is append-only. On restart, the most recent row for each attempt_id
    is used. Successful attempts with an existing local file are skipped.
    Failed attempts are retried when RETRY_FAILED_DOWNLOADS=True.
    """
    state = load_download_state(DOWNLOAD_STATE_JSONL)
    attempt_log = []
    receipts: dict[str, dict] = {}

    session = requests.Session()
    session.headers.update(SESSION.headers)

    for attempt in tqdm(
        attempts,
        desc="Downloading legacy documents",
        unit="file",
        dynamic_ncols=True,
    ):
        aid = attempt["attempt_id"]
        previous = state.get(aid)

        if (
            previous
            and previous.get("status") in {"downloaded", "already_present"}
            and result_path_exists(previous)
            and not OVERWRITE_EXISTING_FILES
        ):
            attempt_log.append(previous)
            receipts[previous["relative_path"]] = previous
            continue

        if (
            previous
            and previous.get("status") == "failed"
            and not RETRY_FAILED_DOWNLOADS
        ):
            attempt_log.append(previous)
            continue

        existing = existing_exact_attempt_path(attempt)
        if existing is not None and not OVERWRITE_EXISTING_FILES:
            result = {
                **attempt,
                "status": "already_present",
                "http_status": None,
                "relative_path": repo_relative(existing),
                "retrieved_at_utc": "",
                "error": "",
            }
            attempt_log.append(result)
            receipts[result["relative_path"]] = result

            # Only add a checkpoint if this is new information.
            if previous != result:
                append_jsonl_row(DOWNLOAD_STATE_JSONL, result)
                state[aid] = result
            continue

        out_path = (
            celex_output_path(
                attempt["celex"],
                attempt["language"],
                attempt["file_format"],
            )
            if attempt["source_type"] == "celex_eurlex"
            else direct_output_path(
                attempt["source_entry_id"],
                attempt["source_url"],
                attempt["file_format"],
            )
        )
        out_path.parent.mkdir(parents=True, exist_ok=True)

        try:
            response = session.get(
                attempt["source_url"],
                timeout=DOCUMENT_REQUEST_TIMEOUT_SECONDS,
                allow_redirects=True,
            )
            content = response.content or b""
            content_type = response.headers.get("content-type", "")

            if (
                response.status_code == 200
                and valid_download(content, attempt["file_format"], content_type)
            ):
                out_path.write_bytes(content)
                result = {
                    **attempt,
                    "status": "downloaded",
                    "http_status": response.status_code,
                    "relative_path": repo_relative(out_path),
                    "retrieved_at_utc": utc_now_iso(),
                    "error": "",
                }
                receipts[result["relative_path"]] = result
            else:
                result = {
                    **attempt,
                    "status": "failed",
                    "http_status": response.status_code,
                    "relative_path": "",
                    "retrieved_at_utc": utc_now_iso(),
                    "error": (
                        f"status={response.status_code}; "
                        f"content_type={content_type}; bytes={len(content)}"
                    ),
                }

        except Exception as exc:
            logger.exception("Document download failed: %s", attempt["source_url"])
            result = {
                **attempt,
                "status": "failed",
                "http_status": None,
                "relative_path": "",
                "retrieved_at_utc": utc_now_iso(),
                "error": repr(exc),
            }

        attempt_log.append(result)
        append_jsonl_row(DOWNLOAD_STATE_JSONL, result)
        state[aid] = result
        time.sleep(DOCUMENT_REQUEST_SLEEP_SECONDS)

    return attempt_log, receipts


download_attempts = build_download_attempts(parsed_records)

if DOWNLOAD_DOCUMENTS:
    download_log, download_receipts = run_document_downloads(download_attempts)
else:
    download_log = list(load_download_state(DOWNLOAD_STATE_JSONL).values())
    download_receipts = {
        row["relative_path"]: row
        for row in download_log
        if row.get("status") in {"downloaded", "already_present"}
        and clean_text(row.get("relative_path", ""))
        and (PROJECT_ROOT / row["relative_path"]).exists()
    }
    print("Document retrieval skipped (DOWNLOAD_DOCUMENTS=False).")

latest_download_state = load_download_state(DOWNLOAD_STATE_JSONL)
download_state_df = pd.DataFrame(latest_download_state.values())

if download_state_df.empty:
    download_failures_df = pd.DataFrame(columns=[
        "source_entry_id",
        "celex",
        "language",
        "file_format",
        "source_url",
        "http_status",
        "error",
    ])
else:
    failure_columns = [
        "source_entry_id",
        "celex",
        "language",
        "file_format",
        "source_url",
        "http_status",
        "error",
    ]
    download_failures_df = download_state_df.loc[
        download_state_df["status"] == "failed",
        failure_columns,
    ].reset_index(drop=True)

print(f"Document download candidates: {len(download_attempts):,}")
if len(download_state_df):
    print("Latest download state:")
    print(download_state_df["status"].value_counts().to_string())
print(f"Failed download attempts currently recorded: {len(download_failures_df):,}")

if len(download_failures_df):
    display(download_failures_df.head(20))
```

## 5. Inventory retained documents

The document manifest contains one row per retained local file and uses repository-relative
paths. SHA-256 is kept because it is useful file-level provenance.

Two practical review views are produced: download attempts that still fail, and source
records for which no retained document is available.


```python
def detect_html_language(path: Path) -> str:
    try:
        html = decode_html_bytes(path.read_bytes()[:300000])
        soup = BeautifulSoup(html, "html.parser")

        html_tag = soup.find("html")
        if html_tag and html_tag.get("lang"):
            value = str(html_tag.get("lang")).split("-")[0].upper()
            if re.fullmatch(r"[A-Z]{2}", value):
                return value

        for meta in soup.find_all("meta"):
            key = str(
                meta.get("http-equiv", "") or meta.get("name", "")
            ).lower()
            if key in {"content-language", "language"}:
                value = clean_text(meta.get("content", "")).split("-")[0].upper()
                if re.fullmatch(r"[A-Z]{2}", value):
                    return value

        match = re.search(r"/legal-content/([A-Za-z]{2})/", html)
        if match:
            return match.group(1).upper()

    except Exception:
        pass

    return "unknown"


def cached_document_is_navigation(path: Path) -> bool:
    name = path.name.lower()

    if name in {
        "search.html",
        "index.html",
        "top.html",
        "previous.html",
        "next.html",
    }:
        return True

    if re.fullmatch(r"(?:19|20)\d{2}\.html", name):
        return True

    return False


def inventory_documents(
    records: list[dict],
    receipts: dict[str, dict],
) -> tuple[pd.DataFrame, pd.DataFrame]:
    celex_to_entries = defaultdict(list)
    direct_basename_to_sources = defaultdict(list)

    for record in records:
        entry_id = record["source_entry_id"]

        for celex in record["parsed"]["celex"]:
            celex_to_entries[celex].append(entry_id)

        for link in record["links"]:
            if link.get("classification") not in {"full_text", "official_journal"}:
                continue

            basename = Path(urlparse(link.get("url", "")).path).name
            if basename:
                direct_basename_to_sources[basename].append({
                    "source_entry_id": entry_id,
                    "source_type": (
                        "full_text"
                        if link.get("classification") == "full_text"
                        else "official_journal"
                    ),
                    "source_url": link.get("url", ""),
                })

    docs = []
    orphans = []

    files = sorted([
        path
        for folder in [HTML_FILES_DIR, PDF_FILES_DIR]
        for path in folder.rglob("*")
        if path.is_file()
    ])

    for path in files:
        if path.stat().st_size <= 0:
            continue

        if cached_document_is_navigation(path):
            orphans.append({
                "relative_path": repo_relative(path),
                "reason": "navigation_or_source_page",
            })
            continue

        rel = repo_relative(path)
        suffix = path.suffix.lower().lstrip(".")
        file_format = (
            "pdf"
            if suffix == "pdf"
            else "html"
            if suffix in {"html", "htm"}
            else suffix
        )

        celex = ""
        source_type = ""
        source_url = ""
        language = "unknown"
        mapped_entries: list[str] = []

        celex_match = re.match(
            r"^(3\d{4}[A-Z]\d{4})(?:_([A-Za-z]{2}))?\.(?:html?|pdf)$",
            path.name,
            flags=re.I,
        )

        if celex_match:
            celex = celex_match.group(1).upper()
            language = (
                celex_match.group(2).upper()
                if celex_match.group(2)
                else "unknown"
            )
            mapped_entries = dedupe_keep_order(celex_to_entries.get(celex, []))
            source_type = "celex_eurlex"

            if language != "unknown":
                source_url = eurlex_download_url(
                    celex,
                    language,
                    file_format,
                )

        else:
            basename = path.name.split("__", 1)[-1]
            sources = direct_basename_to_sources.get(basename, [])
            mapped_entries = dedupe_keep_order(
                [item["source_entry_id"] for item in sources]
            )

            if sources:
                source_type = sources[0]["source_type"]
                source_url = sources[0]["source_url"]
                language = infer_link_language(source_url)

        if not mapped_entries:
            orphans.append({
                "relative_path": rel,
                "reason": "no_source_record_mapping",
            })
            continue

        receipt = receipts.get(rel)
        if receipt:
            language = receipt.get("language", language) or language
            source_url = receipt.get("source_url", source_url)
            source_type = receipt.get("source_type", source_type)
            celex = receipt.get("celex", celex)

        if language == "unknown" and file_format == "html":
            language = detect_html_language(path)

        file_hash = sha256_file(path)
        document_id_hash = hashlib.sha256(
            f"{rel}\0{file_hash}".encode("utf-8")
        ).hexdigest()

        docs.append({
            "document_id": f"COMLEGDOC_{document_id_hash[:16]}",
            "source_entry_id": mapped_entries[0],
            "related_source_entry_ids": "; ".join(mapped_entries),
            "celex": celex,
            "source_type": source_type,
            "language": language,
            "file_format": file_format,
            "source_url": source_url,
            "relative_path": rel,
            "sha256": file_hash,
        })

    document_columns = [
        "document_id",
        "source_entry_id",
        "related_source_entry_ids",
        "celex",
        "source_type",
        "language",
        "file_format",
        "source_url",
        "relative_path",
        "sha256",
    ]

    documents = pd.DataFrame(docs, columns=document_columns)
    orphan_df = pd.DataFrame(
        orphans,
        columns=["relative_path", "reason"],
    )

    return documents, orphan_df


documents_df, orphan_documents_df = inventory_documents(
    parsed_records,
    download_receipts,
)

if len(orphan_documents_df):
    orphan_documents_df.to_csv(
        LOG_DIR / "orphan_cached_documents.csv",
        index=False,
    )

# Count every mapped document for every associated source record.
doc_counts = Counter()
for value in documents_df["related_source_entry_ids"].fillna("").astype(str):
    for entry_id in [item.strip() for item in value.split(";") if item.strip()]:
        doc_counts[entry_id] += 1

if len(records_df):
    records_df["document_count"] = (
        records_df["source_entry_id"]
        .map(doc_counts)
        .fillna(0)
        .astype(int)
    )

missing_documents_df = records_df.loc[
    records_df["document_count"] == 0,
    [
        "source_entry_id",
        "source_year",
        "title",
        "decision_date",
        "case_numbers",
        "celex_ids",
        "source_url",
    ],
].reset_index(drop=True)

print(f"Retained local documents mapped to source records: {len(documents_df):,}")
print(f"Source records with no retained document: {len(missing_documents_df):,}")
print(f"Unmapped/non-document cached files: {len(orphan_documents_df):,}")

if len(missing_documents_df):
    display(missing_documents_df.head(25))
```

## 6. Export concise CSV and Excel views

Outputs:

- `com_legacy_records.csv` — compact source-record table;
- `com_legacy_documents.csv` — compact document manifest;
- `com_legacy.xlsx` — the same two tables plus `missing_documents`,
  `download_failures`, and a small `source_pages` overview.

No audit workbook, validation sheet, output-reconciliation checks, or nested audit JSONL
are generated.


```python
source_pages_export_df = source_pages_df[[
    "source_year",
    "source_url",
    "relative_cache_path",
    "acquisition_status",
    "parse_status",
    "record_count",
]].copy()

records_df.to_csv(
    RECORDS_CSV,
    index=False,
    encoding="utf-8",
)
documents_df.to_csv(
    DOCUMENTS_CSV,
    index=False,
    encoding="utf-8",
)

with pd.ExcelWriter(WORKBOOK_XLSX, engine="openpyxl") as writer:
    records_df.to_excel(writer, sheet_name="records", index=False)
    documents_df.to_excel(writer, sheet_name="documents", index=False)
    missing_documents_df.to_excel(
        writer,
        sheet_name="missing_documents",
        index=False,
    )
    download_failures_df.to_excel(
        writer,
        sheet_name="download_failures",
        index=False,
    )
    source_pages_export_df.to_excel(
        writer,
        sheet_name="source_pages",
        index=False,
    )

# Light formatting only: filters, frozen header and sensible widths.
from openpyxl import load_workbook
from openpyxl.styles import Alignment, Font

wb = load_workbook(WORKBOOK_XLSX)

for ws in wb.worksheets:
    ws.freeze_panes = "A2"

    if ws.max_row >= 1 and ws.max_column >= 1:
        ws.auto_filter.ref = ws.dimensions
        for cell in ws[1]:
            cell.font = Font(bold=True)

    for column_cells in ws.columns:
        letter = column_cells[0].column_letter
        sample = [
            str(cell.value or "")
            for cell in column_cells[: min(len(column_cells), 150)]
        ]
        width = min(
            max(max((len(value) for value in sample), default=8) + 2, 10),
            50,
        )
        ws.column_dimensions[letter].width = width

    for row in ws.iter_rows():
        for cell in row:
            if isinstance(cell.value, str) and len(cell.value) > 80:
                cell.alignment = Alignment(
                    wrap_text=True,
                    vertical="top",
                )

wb.save(WORKBOOK_XLSX)

print("Publication/review outputs:")
print(f"- {repo_relative(RECORDS_CSV)}")
print(f"- {repo_relative(DOCUMENTS_CSV)}")
print(f"- {repo_relative(WORKBOOK_XLSX)}")
print()
print("Operational state:")
print(f"- {repo_relative(DOWNLOAD_STATE_JSONL)}")
```

## 7. Run summary

The final summary is intentionally practical: whether the source pages were available,
how many records/documents were retained, and what still needs attention.


```python
unavailable_years = source_pages_df.loc[
    source_pages_df["parse_status"] == "source_unavailable",
    "source_year",
].astype(int).tolist()

review_years = source_pages_df.loc[
    source_pages_df["parse_status"] == "parser_review_required",
    "source_year",
].astype(int).tolist()

print("Run summary")
print("-----------")
print(f"Source pages: {len(YEARS) - len(unavailable_years)}/{len(YEARS)} available")
print(f"Parsed records: {len(records_df):,}")
print(f"Retained documents: {len(documents_df):,}")
print(f"Records with no retained document: {len(missing_documents_df):,}")
print(f"Failed download attempts: {len(download_failures_df):,}")

if unavailable_years:
    print(f"Unavailable source years: {unavailable_years}")

if review_years:
    print(f"Years requiring parser review: {review_years}")

if len(records_df):
    print(
        f"Records without CELEX: "
        f"{int(records_df['celex_ids'].eq('').sum()):,}"
    )
```

## Reproducibility notes

Required packages: `pandas`, `requests`, `beautifulsoup4`, `tqdm`, and `openpyxl`.

Raw annual pages and downloaded source documents stay under `data/raw/com_legacy/`.
Operational checkpoint/log files stay under `logs/com_legacy/`. Publication/review
tables are written to `output/com_legacy/`.

The notebook does not treat manual review as a machine-generated validation field.
If a missing source document has already been checked manually, that remains part of
the research workflow rather than an additional status column.
