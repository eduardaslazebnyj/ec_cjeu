# 11 — Process the manually edited final Commission manifest

Place `final_commission_manifest_manual_changes.xlsx` in `input/`, then run all cells. Any populated `candidate_to_delete` removes that row. Remaining rows with the same `candidate_to_merge` become one decision group. Every distinct, nonblank value from every column is carried into the consolidated sheet, in original row order, separated by ` | ` when values differ.

The output `output/final_manifest/final_commission_manifest_processed.xlsx` contains `final_manifest` (one row per merge group) and `retained_files` (each surviving source-file row). The second sheet preserves the association between source, downloaded filename, and cleaned filename. A missing merge ID is treated as its own standalone row.


```python
from pathlib import Path
from datetime import date, datetime
import re
import pandas as pd
from openpyxl.styles import Font, PatternFill
from openpyxl.utils import get_column_letter


def project_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for directory in (start, *start.parents):
        if (directory / "input").is_dir():
            return directory
    raise FileNotFoundError("No input/ directory found in this directory or its parents. Set ROOT to your ECCJEU project path.")

ROOT = project_root()  # Or set explicitly: ROOT = Path("/path/to/eccjeu")
INPUT = ROOT / "input/final_commission_manifest_manual_changes.xlsx"
OUTPUT = ROOT / "output/final_manifest/final_commission_manifest_processed.xlsx"
if not INPUT.is_file():
    raise FileNotFoundError(f"Place the manually edited Excel file here: {INPUT}")
print("Input:", INPUT)
print("Output:", OUTPUT)
```

```python
# Read the file-level sheet by name if present; otherwise read the first sheet.
book = pd.ExcelFile(INPUT)
sheet = "files" if "files" in book.sheet_names else book.sheet_names[0]
files = pd.read_excel(INPUT, sheet_name=sheet, dtype=object, keep_default_na=False)
files.columns = [str(column).strip() for column in files.columns]
if files.columns.duplicated().any():
    raise ValueError("Duplicate column names found in the input sheet.")
required = {"candidate_to_merge", "candidate_to_delete"}
missing = required - set(files.columns)
if missing:
    raise ValueError(f"The input sheet {sheet!r} is missing columns: {sorted(missing)}")
print(f"Read {len(files):,} rows from {sheet!r}; {len(files.columns)} columns.")
```

```python
FALSE_MARKERS = {"false", "no", "n", "0", "keep"}


def delete_marked(value):
    if value is None or value is False:
        return False
    if isinstance(value, (int, float)) and not isinstance(value, bool):
        return bool(value) and not pd.isna(value)
    label = str(value).strip()
    return bool(label) and label.casefold() not in FALSE_MARKERS


def merge_id(value, excel_row):
    if value is None or str(value).strip() == "":
        return f"standalone_row_{excel_row}"
    if isinstance(value, (int, float)) and not isinstance(value, bool):
        if pd.isna(value):
            return f"standalone_row_{excel_row}"
        if float(value).is_integer():
            return str(int(value))
    label = str(value).strip()
    if re.fullmatch(r"\d+\.0+", label):
        return label.split(".")[0]
    return label


retained = files.loc[~files["candidate_to_delete"].map(delete_marked)].copy()
retained["candidate_to_delete"] = ""
# Keep the original Excel row number for standalone IDs and auditability.
retained.insert(0, "input_excel_row", retained.index + 2)
retained["candidate_to_merge"] = [merge_id(value, int(row)) for value, row in zip(
    retained["candidate_to_merge"], retained["input_excel_row"]
)]
removed_count = len(files) - len(retained)
print(f"Deleted {removed_count:,} marked rows; retained {len(retained):,} rows.")
```

```python
SEPARATOR = " | "


def display_value(value):
    if value is None or (not isinstance(value, (list, dict, set)) and pd.isna(value)):
        return ""
    if isinstance(value, (datetime, date, pd.Timestamp)):
        return value.strftime("%Y-%m-%d")
    if isinstance(value, float) and value.is_integer():
        return str(int(value))
    return str(value).strip()


def combine_values(values):
    distinct = []
    seen = set()
    for value in values:
        rendered = display_value(value)
        if rendered and rendered not in seen:
            distinct.append(rendered)
            seen.add(rendered)
    result = SEPARATOR.join(distinct)
    if len(result) > 32767:
        raise ValueError("A merged Excel cell exceeds the 32,767-character limit. Inspect its group in retained_files.")
    return result


# Aggregate every input column, including source-specific paths, filenames, and classifications.
# A separate file-level sheet preserves which values belong to which source row.
columns = list(files.columns)
grouped = retained.groupby("candidate_to_merge", sort=False, dropna=False)
records = []
for group_id, group in grouped:
    record = {"candidate_to_merge": group_id, "candidate_to_delete": ""}
    for column in columns:
        if column not in ("candidate_to_merge", "candidate_to_delete"):
            try:
                record[column] = combine_values(group[column])
            except ValueError as exc:
                raise ValueError(f"Merge ID {group_id!r}, column {column!r}: {exc}") from exc
    record["merged_file_count"] = len(group)
    records.append(record)

final = pd.DataFrame.from_records(records, columns=columns + ["merged_file_count"])
# Sort by earliest case date, with merge ID breaking ties. Unparseable dates sort last.
date_column = next((c for c in ("case_date", "decision_date", "date") if c in retained.columns), None)
if date_column and len(final):
    earliest = retained.assign(_parsed_date=pd.to_datetime(retained[date_column], errors="coerce", dayfirst=False))
    minima = earliest.groupby("candidate_to_merge", sort=False)["_parsed_date"].min()
    final["_sort_date"] = final["candidate_to_merge"].map(minima)
    final = final.sort_values(["_sort_date", "candidate_to_merge"], kind="stable", na_position="last").drop(columns="_sort_date")
final = final.reset_index(drop=True)
print(f"Created {len(final):,} merged decisions from {len(retained):,} retained file rows.")
print("Groups with multiple retained files:", int((final["merged_file_count"] > 1).sum()) if len(final) else 0)
```

```python
OUTPUT.parent.mkdir(parents=True, exist_ok=True)
pending = OUTPUT.with_name(OUTPUT.stem + ".pending.xlsx")
with pd.ExcelWriter(pending, engine="openpyxl") as writer:
    final.to_excel(writer, sheet_name="final_manifest", index=False)
    retained.to_excel(writer, sheet_name="retained_files", index=False)
    for name in ("final_manifest", "retained_files"):
        ws = writer.sheets[name]
        ws.freeze_panes = "C2" if name == "final_manifest" else "D2"
        ws.auto_filter.ref = ws.dimensions
        for cell in ws[1]:
            cell.fill = PatternFill("solid", fgColor="19334D")
            cell.font = Font(color="FFFFFF", bold=True)
        for index, column in enumerate((final if name == "final_manifest" else retained).columns, start=1):
            ws.column_dimensions[get_column_letter(index)].width = min(65, max(17, len(str(column)) + 3))
pending.replace(OUTPUT)
print("Wrote:", OUTPUT)
```
