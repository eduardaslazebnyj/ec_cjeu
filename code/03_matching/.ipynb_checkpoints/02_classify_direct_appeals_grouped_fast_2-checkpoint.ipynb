# 02 — Classify direct appeals to Commission decisions

This notebook classifies the relationship between EU Court documents and the Commission decision identifiers detected in `01_court_decision_regex_v2`.

The model-call unit is now **one Court document**, not one Court document × one Commission identifier. All still-pending Commission identifiers detected in the same Court document are classified together in a single request. The output is then expanded back to the original **document × identifier** level so downstream files keep the same structure.

The main question is whether the present Court proceeding is a **direct judicial challenge to each identified Commission decision**. If it is not, the notebook assigns a broad citation role describing how that Commission decision is used in the Court document.

This notebook does not search for additional Commission decisions. Candidate discovery belongs to notebook 01; this notebook only classifies the candidates already identified there.


## Inputs and outputs

**Input**

- `output/court_regex/court_decision_regex.xlsx`
- sheet `references`, produced by `01_court_decision_regex_v2`

Only the two Commission-decision identifier families used for relationship classification are retained:

- `decision_case_number`
- `decision_oj_reference`

**Prepared model inputs**

Classification is performed once per Court document. For each model call, the prompt contains:

- the first `BEGINNING_CHARS` characters (procedural header, parties, subject-matter);
- the last `ENDING_CHARS` characters (dispositif / conclusion);
- a ±`HIT_CONTEXT_CHARS` window around the **first** occurrence of every pending identifier;
- additional ±`HIT_CONTEXT_CHARS` windows for every subsequent occurrence, added greedily while the total stays within `MAX_PREPARED_DOCUMENT_CHARS`;
- overlapping windows merged.

If the mandatory content (anchors + first-hit windows) alone would exceed `MAX_PREPARED_DOCUMENT_CHARS`, the document is **deferred** (`classification_status = "deferred"`) rather than sending an oversized or incomplete prompt. Deferred records are written to the checkpoint and skipped on reruns until handled separately.

**Runtime log**

- `logs/court_regex/direct_appeals_classification.jsonl`

The JSONL remains candidate-level even though model calls are document-level. Each successful model response is split into one checkpoint record per identifier, keyed by `job_id` (sha1 of `record_id|reference_type|reference_normalized`). Existing successful and deferred records survive reruns unchanged.

**Final output**

- `output/court_regex/court_direct_appeals.xlsx`

The final workbook remains at the original document × Commission-identifier level.

```python
from __future__ import annotations

from datetime import datetime, timezone
from pathlib import Path
from typing import Any
import asyncio
import hashlib
import json
import os
import re

import httpx
import pandas as pd
from tqdm.auto import tqdm
```

## 1. Paths and settings

```python
def find_project_root(start: Path | None = None) -> Path:
    override = os.environ.get("ECCJEU_PROJECT_ROOT", "").strip()
    if override:
        root = Path(override).expanduser().resolve()
        if not root.exists():
            raise FileNotFoundError(f"ECCJEU_PROJECT_ROOT does not exist: {root}")
        return root

    start = (start or Path.cwd()).resolve()

    for candidate in [start, *start.parents]:
        if (candidate / "code" / "03_matching").exists():
            return candidate

    for candidate in [start, *start.parents]:
        if (candidate / ".git").exists():
            return candidate

    for candidate in [start, *start.parents]:
        if (candidate / "output").exists():
            return candidate

    raise FileNotFoundError(
        "Could not locate the project root. Run the notebook from within the "
        "repository or set ECCJEU_PROJECT_ROOT."
    )


PROJECT_ROOT = find_project_root()

REGEX_XLSX = PROJECT_ROOT / "output" / "court_regex" / "court_decision_regex.xlsx"
OUTPUT_DIR = PROJECT_ROOT / "output" / "court_regex"
LOG_DIR = PROJECT_ROOT / "logs" / "court_regex"
CHECKPOINT_JSONL = LOG_DIR / "direct_appeals_classification.jsonl"
OUTPUT_XLSX = OUTPUT_DIR / "court_direct_appeals.xlsx"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
LOG_DIR.mkdir(parents=True, exist_ok=True)

# Prepared-input strategy
# One model request is made per Court document. The focused input contains:
#   - the first BEGINNING_CHARS characters (procedural opening, parties);
#   - the last ENDING_CHARS characters (dispositif / conclusion);
#   - a ±HIT_CONTEXT_CHARS window around EVERY regex occurrence of every target;
#   - overlapping windows merged.
# Anchor sections (beginning + ending) are always kept. Hit windows are added
# greedily in occurrence order and dropped once MAX_PREPARED_DOCUMENT_CHARS is reached.
BEGINNING_CHARS = 5_000
ENDING_CHARS = 3_000
HIT_CONTEXT_CHARS = 1_000
MAX_PREPARED_DOCUMENT_CHARS = 30_000

# Local vLLM / OpenAI-compatible endpoint
VLLM_BASE_URL = os.getenv("VLLM_BASE_URL", "http://localhost:8000/v1")
VLLM_API_KEY = os.getenv("VLLM_API_KEY", "EMPTY")
MODEL = os.getenv("VLLM_MODEL", "google/gemma-4-31B-it")

# Start conservatively. Increase to 6 or 8 after confirming the local server is stable.
MAX_WORKERS = int(os.getenv("DIRECT_APPEAL_WORKERS", "8"))
REQUEST_TIMEOUT_SECONDS = 3600
MAX_RETRIES = 2
TEMPERATURE = 0.0
MAX_OUTPUT_TOKENS_PER_TARGET = 120
MAX_OUTPUT_TOKENS_BASE = 160
MAX_OUTPUT_TOKENS_CAP = 2200

# "update" resumes from successful JSONL records.
# "redo_all" ignores the existing checkpoint and reruns every candidate.
# "rebuild_excel" makes no model calls and only rebuilds the workbook.
RUN_MODE = os.getenv("RUN_MODE", "update").strip().lower()
if RUN_MODE not in {"update", "redo_all", "rebuild_excel"}:
    raise ValueError("RUN_MODE must be update, redo_all, or rebuild_excel")

print("PROJECT_ROOT:", PROJECT_ROOT)
print("Regex input:", REGEX_XLSX)
print("Checkpoint:", CHECKPOINT_JSONL)
print("Output:", OUTPUT_XLSX)
print("Workers:", MAX_WORKERS)
print("Run mode:", RUN_MODE)
```

## 2. Load the aggregated Commission-decision candidates

Notebook 01 already groups repeated mentions of the same normalized identifier within a Court document and stores the position of every underlying regex hit in `hit_spans_json`.

Those rows remain the candidate-level output unit. For model inference, however, candidates are grouped by `record_id` so that all identifiers from the same Court document can be classified in one request.


```python
if not REGEX_XLSX.exists():
    raise FileNotFoundError(
        f"Required regex workbook is missing: {REGEX_XLSX}\n"
        "Run 01_court_decision_regex_v2 first."
    )

references = pd.read_excel(
    REGEX_XLSX,
    sheet_name="references",
    dtype=str,
    keep_default_na=False,
)
references.columns = [str(c).strip() for c in references.columns]

required_columns = {
    "record_id",
    "affId",
    "procedureId",
    "case_number_clean",
    "clean_file_path",
    "reference_type",
    "reference_normalized",
    "reference_count",
    "hit_spans_json",
}
missing = sorted(required_columns - set(references.columns))
if missing:
    raise ValueError(f"Regex references sheet is missing required columns: {missing}")

CLASSIFIED_REFERENCE_TYPES = {"decision_case_number", "decision_oj_reference"}

jobs_df = references.loc[
    references["reference_type"].isin(CLASSIFIED_REFERENCE_TYPES)
].copy()

jobs_df = jobs_df.reset_index(drop=True)

def stable_job_id(row: pd.Series) -> str:
    key = "|".join(
        [
            str(row.get("record_id", "")),
            str(row.get("reference_type", "")),
            str(row.get("reference_normalized", "")),
        ]
    )
    return hashlib.sha1(key.encode("utf-8")).hexdigest()[:16]

jobs_df["job_id"] = jobs_df.apply(stable_job_id, axis=1)

if jobs_df["job_id"].duplicated().any():
    dupes = jobs_df.loc[jobs_df["job_id"].duplicated(keep=False)]
    raise ValueError(
        "Duplicate classification job IDs were produced. Inspect:\n"
        + dupes[["record_id", "reference_type", "reference_normalized"]]
        .head(20)
        .to_string(index=False)
    )

print(f"Aggregated Commission-decision candidates: {len(jobs_df):,}")
display(
    jobs_df["reference_type"]
    .value_counts()
    .rename_axis("reference_type")
    .reset_index(name="candidates")
)
display(jobs_df.head(20))
```

## 3. Prepare document-level model inputs

Model inputs are assembled at runtime for each Court document from its still-pending identifiers.

The focused input is built in four priority tiers:

1. **Beginning anchor** — the first `BEGINNING_CHARS` characters (procedural header, parties, subject-matter). Always included.
2. **Ending anchor** — the last `ENDING_CHARS` characters (dispositif, operative part). Always included.
3. **Guaranteed first-hit contexts** — a ±`HIT_CONTEXT_CHARS` window around the first regex occurrence of every pending identifier. Always included.
4. **Extra-hit contexts** — a ±`HIT_CONTEXT_CHARS` window around every subsequent occurrence of every identifier. Added greedily; dropped once the budget is reached.

Overlapping windows are merged before the text is assembled. `MAX_PREPARED_DOCUMENT_CHARS` is a **hard cap**: if tiers 1–3 alone would exceed it, the document is **deferred** (`classification_status = "deferred"`) rather than sending an oversized prompt. In practice this only affects documents with an unusually large number of distinct identifiers.

```python
ILLEGAL_MODEL_CHARS_RE = re.compile(r"[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]")


class BudgetExceededError(Exception):
    """Raised when anchors + guaranteed first-hit contexts exceed MAX_PREPARED_DOCUMENT_CHARS."""


def resolve_repo_path(value: str) -> Path | None:
    raw = str(value or "").strip()
    if not raw:
        return None

    p = Path(raw).expanduser()
    if p.is_absolute():
        return p.resolve()

    return (PROJECT_ROOT / p).resolve()


def clean_model_text(text: str) -> str:
    """Remove problematic controls while preserving ordinary line structure."""
    text = str(text or "")
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = ILLEGAL_MODEL_CHARS_RE.sub("", text)
    return text.encode("utf-8", errors="replace").decode("utf-8")


def parse_hit_spans(value: str) -> list[tuple[int, int]]:
    try:
        raw = json.loads(str(value or "[]"))
    except json.JSONDecodeError as exc:
        raise ValueError(f"Invalid hit_spans_json: {value!r}") from exc

    spans: list[tuple[int, int]] = []
    for item in raw:
        if not isinstance(item, list) or len(item) != 2:
            raise ValueError(f"Invalid hit span: {item!r}")
        start, end = int(item[0]), int(item[1])
        if start < 0 or end < start:
            raise ValueError(f"Invalid hit span boundaries: {item!r}")
        spans.append((start, end))
    return spans


def merge_intervals(
    intervals: list[tuple[int, int]],
    *,
    text_length: int,
) -> list[tuple[int, int]]:
    clipped = sorted(
        (max(0, int(start)), min(text_length, int(end)))
        for start, end in intervals
        if int(end) > 0 and int(start) < text_length
    )
    if not clipped:
        return []

    merged: list[list[int]] = [[clipped[0][0], clipped[0][1]]]
    for start, end in clipped[1:]:
        last = merged[-1]
        if start <= last[1]:
            last[1] = max(last[1], end)
        else:
            merged.append([start, end])
    return [(start, end) for start, end in merged]


def build_document_input(
    source_text: str,
    target_rows: list[dict[str, Any]],
) -> tuple[str, list[tuple[int, int]]]:
    """Build one focused text input for all pending identifiers in a Court document.

    Priority order (highest to lowest):
      1. Beginning anchor  — first BEGINNING_CHARS characters, always kept.
      2. Ending anchor     — last ENDING_CHARS characters, always kept.
      3. First-hit context — ±HIT_CONTEXT_CHARS around the first occurrence of
                             every target identifier, always kept.
      4. Extra-hit contexts — ±HIT_CONTEXT_CHARS around every subsequent
                             occurrence, added greedily while within budget.

    If the combined size of tiers 1–3 already exceeds MAX_PREPARED_DOCUMENT_CHARS,
    a BudgetExceededError is raised and the document is deferred rather than sending
    an oversized or incomplete prompt.
    """
    text = clean_model_text(source_text)
    text_length = len(text)

    def hit_window(start: int, end: int) -> tuple[int, int]:
        return (
            max(0, start - HIT_CONTEXT_CHARS),
            min(text_length, end + HIT_CONTEXT_CHARS),
        )

    # Tiers 1 + 2: anchors.
    required: list[tuple[int, int]] = [
        (0, min(BEGINNING_CHARS, text_length)),
        (max(0, text_length - ENDING_CHARS), text_length),
    ]

    extra_hit_intervals: list[tuple[int, int]] = []

    for row in target_rows:
        spans = parse_hit_spans(row["hit_spans_json"])
        if not spans:
            raise ValueError(
                f"Candidate {row['job_id']} has no stored hit positions"
            )

        for start, end in spans:
            if start > text_length or end > text_length:
                raise ValueError(
                    f"Stored regex span [{start}, {end}] exceeds source-text length "
                    f"{text_length:,}. Rerun notebook 01 if the clean text changed."
                )

        # Tier 3: first hit of each target — mandatory.
        required.append(hit_window(*spans[0]))

        # Tier 4: remaining hits — budget-permitting.
        for span in spans[1:]:
            extra_hit_intervals.append(hit_window(*span))

    # Hard cap check: if mandatory content alone exceeds the budget, defer.
    required_merged = merge_intervals(required, text_length=text_length)
    required_chars = sum(end - start for start, end in required_merged)
    if required_chars > MAX_PREPARED_DOCUMENT_CHARS:
        raise BudgetExceededError(
            f"Required input ({required_chars:,} chars) exceeds "
            f"MAX_PREPARED_DOCUMENT_CHARS ({MAX_PREPARED_DOCUMENT_CHARS:,}) "
            f"for {len(target_rows)} targets. Deferring this document."
        )

    # Tier 4: greedily add extra hit windows while under the hard cap.
    selected = required_merged
    for interval in extra_hit_intervals:
        candidate = merge_intervals(selected + [interval], text_length=text_length)
        if sum(end - start for start, end in candidate) <= MAX_PREPARED_DOCUMENT_CHARS:
            selected = candidate

    sections: list[str] = [
        (
            "The extracts below contain the beginning and ending of the Court document "
            "plus local context around every detected Commission identifier. "
            "Overlapping windows have been merged."
        ),
    ]

    for index, (start, end) in enumerate(selected, start=1):
        sections.extend(
            [
                "",
                f"===== SOURCE CHARACTERS {start}:{end} | SECTION {index}/{len(selected)} =====",
                text[start:end],
            ]
        )

    return clean_model_text("\n".join(sections)), selected


def document_group_id(record_id: str) -> str:
    return hashlib.sha1(str(record_id).encode("utf-8")).hexdigest()[:16]


print(
    f"Candidate rows: {len(jobs_df):,} across "
    f"{jobs_df['record_id'].nunique():,} Court documents"
)
```

## 4. Classification taxonomy

The first decision is whether the present Court proceeding is a **direct challenge to the identified Commission decision**.

`is_direct_appeal = true` means that the Commission decision itself is the measure directly challenged in this Court proceeding.

This deliberately does **not** include a later Court of Justice proceeding that directly appeals a General Court judgment, even when that General Court judgment concerned the same Commission decision. Such broader appeal-chain relationships can be added in a later step.

When the Commission decision is not directly challenged, one overall citation role is assigned:

- `legal_substantive` — the decision is used for legal reasoning, precedent, comparison, interpretation, or substantive argument;
- `procedural_evidentiary` — the decision is discussed mainly in relation to procedure, investigation, evidence, access, admissibility, or another procedural/evidentiary point;
- `background_historical` — the decision mainly forms part of the factual, institutional, chronological, or related-case background;
- `remedy_fine_penalty` — the decision is discussed mainly in relation to fines, penalties, remedies, sanctions, or their calculation;
- `unclear` — the supplied text does not support a reliable overall classification.

The classification is made for the **candidate as a whole**, using all supplied mentions together. It is not a majority vote over individual regex hits.

```python
SYSTEM_MESSAGE = """
You are classifying relationships in a legal research dataset.

Use only the supplied Court-document text. Do not use facts from memory.
The target Commission identifiers have already been detected by an earlier regex step.
Classify every supplied target identifier separately, but do so from this one Court
document in a single response.

Return only JSON matching the requested schema.
""".strip()

CITATION_ROLES = [
    "direct_appeal",
    "legal_substantive",
    "procedural_evidentiary",
    "background_historical",
    "remedy_fine_penalty",
    "unclear",
]

RESULT_ITEM_SCHEMA = {
    "type": "object",
    "properties": {
        "job_id": {"type": "string"},
        "is_direct_appeal": {"type": ["boolean", "null"]},
        "citation_role": {"type": "string", "enum": CITATION_ROLES},
        "reason": {"type": "string"},
        "evidence": {"type": "string"},
    },
    "required": [
        "job_id",
        "is_direct_appeal",
        "citation_role",
        "reason",
        "evidence",
    ],
    "additionalProperties": False,
}

RESPONSE_JSON_SCHEMA = {
    "type": "object",
    "properties": {
        "results": {
            "type": "array",
            "items": RESULT_ITEM_SCHEMA,
        }
    },
    "required": ["results"],
    "additionalProperties": False,
}

CLASSIFICATION_INSTRUCTIONS = """
For EACH target, determine whether the PRESENT COURT PROCEEDING directly
challenges the Commission decision represented by that target.

Use is_direct_appeal = true only when that Commission decision itself is the
measure directly challenged in this Court proceeding. Strong evidence includes
language such as an action or application for annulment of the Commission
decision, a request to annul that decision, or a direct challenge to the fine
or obligations imposed by that decision.

Use is_direct_appeal = false when the target decision is mentioned only as
background, precedent, comparison, procedural history, evidence, a related
matter, or another citation.

Important: if the present document concerns an appeal against a General Court
judgment, and that General Court judgment had reviewed a Commission decision,
do NOT call the present Court of Justice proceeding a direct appeal of the
Commission decision merely because the Commission decision appears in the
history. The direct object of that later appeal is the lower-court judgment.

If is_direct_appeal is true, set citation_role to "direct_appeal".

If is_direct_appeal is false, choose the ONE role that best describes how that
Commission decision functions in the Court document overall:
  legal_substantive        — used for legal reasoning, precedent, comparison,
                             interpretation, or substantive argument
  procedural_evidentiary   — discussed mainly regarding procedure, evidence,
                             access, admissibility, or investigation
  background_historical    — mainly factual, institutional, chronological, or
                             related-case background
  remedy_fine_penalty      — discussed mainly regarding fines, penalties,
                             remedies, sanctions, or their calculation
  unclear                  — the supplied text does not support a reliable
                             overall classification

If the direct-appeal question cannot be determined reliably from the supplied
extracts, set is_direct_appeal to null and citation_role to "unclear".

Return exactly one result object for every supplied job_id.
Keep reason to one short sentence. Evidence must be a very short phrase from
the supplied text, not a long quotation.
""".strip()


def build_document_prompt(
    *,
    court_case: str,
    targets: list[dict[str, Any]],
    document_text: str,
) -> str:
    target_parts = []
    for row in targets:
        target_parts.append(
            f"{row['reference_normalized']} ({row['reference_type']})"
            f"  [job_id={row['job_id']}]"
        )
    targets_block = " | ".join(target_parts)

    return (
        f"PRESENT COURT CASE: {court_case or '[not available]'}"
        f"\nTARGETS: {targets_block}"
        f"\n\n{CLASSIFICATION_INSTRUCTIONS}"
        f"\n\nCOURT DOCUMENT EXTRACTS\n"
        f"{'─' * 40}\n"
        f"{document_text}"
    )
```

## 5. Resumable document-level classification

Pending candidate rows are grouped by `record_id`. Each Court document is sent to
the model exactly once, with all of its still-pending identifiers listed together.

After a successful response, results are expanded into one JSONL record per
identifier, keyed by `job_id`. On reruns, already-successful identifiers are
skipped even if other identifiers from the same document are retried. The 36
records already written by the previous notebook are preserved automatically
because the `job_id` hash function is unchanged.

```python
def utc_now_iso() -> str:
    return datetime.now(timezone.utc).isoformat()


def append_jsonl(path: Path, record: dict[str, Any]) -> None:
    with path.open("a", encoding="utf-8") as handle:
        handle.write(json.dumps(record, ensure_ascii=False) + "\n")


def load_latest_checkpoint(path: Path) -> dict[str, dict[str, Any]]:
    latest: dict[str, dict[str, Any]] = {}
    if not path.exists():
        return latest

    with path.open("r", encoding="utf-8", errors="replace") as handle:
        for line_number, line in enumerate(handle, start=1):
            line = line.strip()
            if not line:
                continue
            try:
                record = json.loads(line)
            except json.JSONDecodeError:
                print(f"Skipping malformed checkpoint line {line_number}")
                continue

            job_id = str(record.get("job_id", "")).strip()
            if job_id:
                latest[job_id] = record

    return latest


async def call_vllm(
    client: httpx.AsyncClient,
    prompt: str,
    *,
    target_count: int,
) -> dict[str, Any]:
    endpoint = f"{VLLM_BASE_URL.rstrip('/')}/chat/completions"
    max_tokens = min(
        MAX_OUTPUT_TOKENS_CAP,
        MAX_OUTPUT_TOKENS_BASE + MAX_OUTPUT_TOKENS_PER_TARGET * target_count,
    )
    payload = {
        "model": MODEL,
        "messages": [
            {"role": "system", "content": SYSTEM_MESSAGE},
            {"role": "user", "content": prompt},
        ],
        "temperature": TEMPERATURE,
        "max_tokens": max_tokens,
        "structured_outputs": {"json": RESPONSE_JSON_SCHEMA},
    }
    headers = {"Authorization": f"Bearer {VLLM_API_KEY}"}

    response = await client.post(endpoint, json=payload, headers=headers)
    response.raise_for_status()

    body = response.json()
    choices = body.get("choices") or []
    if not choices:
        raise ValueError("vLLM response contains no choices")

    content = choices[0].get("message", {}).get("content")
    if not content:
        raise ValueError("vLLM response contains no message content")

    return {
        "content": str(content),
        "finish_reason": str(choices[0].get("finish_reason") or ""),
        "max_tokens": max_tokens,
    }


def validate_group_response(
    parsed: dict[str, Any],
    target_rows: list[dict[str, Any]],
) -> dict[str, dict[str, Any]]:
    results = parsed.get("results")
    if not isinstance(results, list):
        raise ValueError("Response does not contain a results list")

    expected = [str(row["job_id"]) for row in target_rows]
    by_id: dict[str, dict[str, Any]] = {}

    for item in results:
        if not isinstance(item, dict):
            raise ValueError("Response contains a non-object result")
        job_id = str(item.get("job_id", "")).strip()
        if not job_id:
            raise ValueError("Response result is missing job_id")
        if job_id in by_id:
            raise ValueError(f"Duplicate result for job_id {job_id}")
        by_id[job_id] = item

    if set(by_id) != set(expected):
        missing = sorted(set(expected) - set(by_id))
        extra = sorted(set(by_id) - set(expected))
        raise ValueError(
            f"Response job_ids do not match request. Missing={missing[:10]}, "
            f"extra={extra[:10]}"
        )

    for job_id, item in by_id.items():
        if item.get("is_direct_appeal") is True:
            item["citation_role"] = "direct_appeal"
        elif item.get("is_direct_appeal") is None:
            item["citation_role"] = "unclear"
        elif item.get("citation_role") == "direct_appeal":
            raise ValueError(
                f"{job_id}: citation_role cannot be direct_appeal when "
                "is_direct_appeal is false"
            )

    return by_id


async def classify_document(
    target_rows: list[dict[str, Any]],
    client: httpx.AsyncClient,
    semaphore: asyncio.Semaphore,
) -> list[dict[str, Any]]:
    first = target_rows[0]

    source_paths = {str(row.get("clean_file_path", "")).strip() for row in target_rows}
    if len(source_paths) != 1:
        raise ValueError(
            f"record_id {first['record_id']} maps to multiple clean_file_path values: "
            f"{sorted(source_paths)}"
        )

    source_path = resolve_repo_path(first["clean_file_path"])
    group_id = document_group_id(first["record_id"])

    def base_record(row: dict[str, Any]) -> dict[str, Any]:
        return {
            "job_id": row["job_id"],
            "document_group_id": group_id,
            "record_id": row["record_id"],
            "affId": row["affId"],
            "procedureId": row["procedureId"],
            "case_number_clean": row["case_number_clean"],
            "reference_type": row["reference_type"],
            "reference_normalized": row["reference_normalized"],
            "reference_count": row["reference_count"],
            "source_text_path": str(source_path or ""),
            "model": MODEL,
        }

    try:
        if source_path is None or not source_path.exists():
            raise FileNotFoundError(f"Clean Court text is missing: {source_path}")

        source_text = source_path.read_text(encoding="utf-8", errors="replace")
        document_text, sections = build_document_input(source_text, target_rows)
        prompt = build_document_prompt(
            court_case=str(first.get("case_number_clean", "")),
            targets=target_rows,
            document_text=document_text,
        )
    except BudgetExceededError as exc:
        # The mandatory input (anchors + one first-hit per target) already exceeds
        # MAX_PREPARED_DOCUMENT_CHARS. Write a deferred record for every target and
        # move on. These are not retried automatically; handle separately if needed.
        return [
            {
                **base_record(row),
                "classification_status": "deferred",
                "classification_error": str(exc),
                "classified_at": utc_now_iso(),
                "finish_reason": "",
                "raw_response": "",
                "parsed_response": None,
                "document_target_count": len(target_rows),
            }
            for row in target_rows
        ]
    except Exception as exc:
        error = f"{type(exc).__name__}: {exc}"
        return [
            {
                **base_record(row),
                "classification_status": "error",
                "classification_error": error,
                "classified_at": utc_now_iso(),
                "finish_reason": "",
                "raw_response": "",
                "parsed_response": None,
            }
            for row in target_rows
        ]

    last_error = ""
    async with semaphore:
        for attempt in range(1, MAX_RETRIES + 2):
            try:
                result = await call_vllm(
                    client,
                    prompt,
                    target_count=len(target_rows),
                )
                parsed = json.loads(result["content"])
                by_id = validate_group_response(parsed, target_rows)

                records: list[dict[str, Any]] = []
                for row in target_rows:
                    item = by_id[str(row["job_id"])]
                    records.append(
                        {
                            **base_record(row),
                            "classification_status": "success",
                            "classification_error": "",
                            "classified_at": utc_now_iso(),
                            "attempt": attempt,
                            "finish_reason": result["finish_reason"],
                            "raw_response": json.dumps(item, ensure_ascii=False),
                            "parsed_response": item,
                            "document_target_count": len(target_rows),
                            "prepared_input_chars": len(document_text),
                            "prepared_sections_json": json.dumps(sections),
                            "max_output_tokens": result["max_tokens"],
                        }
                    )
                return records

            except Exception as exc:
                last_error = f"{type(exc).__name__}: {exc}"

    return [
        {
            **base_record(row),
            "classification_status": "error",
            "classification_error": last_error,
            "classified_at": utc_now_iso(),
            "attempt": MAX_RETRIES + 1,
            "finish_reason": "",
            "raw_response": "",
            "parsed_response": None,
            "document_target_count": len(target_rows),
            "prepared_input_chars": len(document_text),
            "prepared_sections_json": json.dumps(sections),
        }
        for row in target_rows
    ]


async def run_document_groups(
    pending_df: pd.DataFrame,
    *,
    description: str,
) -> list[dict[str, Any]]:
    if pending_df.empty:
        print(f"{description}: nothing to run")
        return []

    grouped_rows = [
        group.to_dict("records")
        for _, group in pending_df.groupby("record_id", sort=False)
    ]

    semaphore = asyncio.Semaphore(MAX_WORKERS)
    timeout = httpx.Timeout(REQUEST_TIMEOUT_SECONDS)

    print(
        f"{description}: {len(pending_df):,} candidate identifiers in "
        f"{len(grouped_rows):,} Court-document requests"
    )

    async with httpx.AsyncClient(timeout=timeout) as client:
        tasks = [
            asyncio.create_task(
                classify_document(rows, client, semaphore)
            )
            for rows in grouped_rows
        ]

        all_records: list[dict[str, Any]] = []
        for task in tqdm(
            asyncio.as_completed(tasks),
            total=len(tasks),
            desc=description,
        ):
            records = await task
            for record in records:
                append_jsonl(CHECKPOINT_JSONL, record)
            all_records.extend(records)

    return all_records
```

```python
if RUN_MODE == "redo_all" and CHECKPOINT_JSONL.exists():
    CHECKPOINT_JSONL.unlink()

latest_before_run = load_latest_checkpoint(CHECKPOINT_JSONL)

def needs_classification(job_id: str) -> bool:
    if RUN_MODE == "rebuild_excel":
        return False

    previous = latest_before_run.get(str(job_id))
    if previous is None:
        return True

    return previous.get("classification_status") not in {"success", "deferred"}


pending = jobs_df.loc[
    jobs_df["job_id"].map(needs_classification)
].copy()

already_successful = len(jobs_df) - len(pending)
pending_documents = pending["record_id"].nunique()

print(f"Already successful candidate identifiers: {already_successful:,}")
print(f"Pending candidate identifiers: {len(pending):,}")
print(f"Pending Court-document model requests: {pending_documents:,}")

if len(pending):
    display(
        pending.groupby("record_id")
        .size()
        .describe(percentiles=[0.5, 0.9, 0.99])
        .rename("pending_identifiers_per_document")
        .to_frame()
    )
```

```python
# Run one model request per Court document.
# In Jupyter, top-level await is supported.

if RUN_MODE != "rebuild_excel":
    _document_results = await run_document_groups(
        pending,
        description="Classifying Court documents",
    )
```

## 6. Assemble the candidate-level classification table

Although inference is document-level, the checkpoint and final workbook remain at the original candidate level: one row per normalized Commission identifier in one Court document.


```python
latest = load_latest_checkpoint(CHECKPOINT_JSONL)

classification_rows: list[dict[str, Any]] = []

for job_id, record in latest.items():
    parsed = record.get("parsed_response")
    parsed = parsed if isinstance(parsed, dict) else {}

    direct_value = parsed.get("is_direct_appeal")
    if direct_value is True:
        direct_status = "direct_appeal"
    elif direct_value is False:
        direct_status = "citation_only"
    else:
        direct_status = "unclear"

    classification_rows.append(
        {
            "job_id": job_id,
            "classification_status": record.get("classification_status", ""),
            "classification_error": record.get("classification_error", ""),
            "is_direct_appeal": direct_value,
            "direct_appeal_status": direct_status,
            "citation_role": parsed.get("citation_role", ""),
            "classification_reason": parsed.get("reason", ""),
            "classification_evidence": parsed.get("evidence", ""),
            "model": record.get("model", ""),
            "classified_at": record.get("classified_at", ""),
            "document_group_id": record.get("document_group_id", ""),
            "document_target_count": record.get("document_target_count", ""),
            "prepared_input_chars": record.get("prepared_input_chars", ""),
            "source_text_path": record.get("source_text_path", ""),
        }
    )

latest_df = pd.DataFrame(classification_rows)

final = jobs_df.copy()

if len(latest_df):
    final = final.merge(
        latest_df,
        on="job_id",
        how="left",
        validate="one_to_one",
    )
else:
    for col in [
        "classification_status",
        "classification_error",
        "is_direct_appeal",
        "direct_appeal_status",
        "citation_role",
        "classification_reason",
        "classification_evidence",
        "model",
        "classified_at",
        "document_group_id",
        "document_target_count",
        "prepared_input_chars",
        "source_text_path",
    ]:
        final[col] = ""

final["classification_status"] = final["classification_status"].fillna("not_run")
final["direct_appeal_status"] = final["direct_appeal_status"].fillna("")

summary = (
    final.groupby(
        ["reference_type", "direct_appeal_status", "citation_role"],
        dropna=False,
    )
    .size()
    .reset_index(name="rows")
    .sort_values(
        ["reference_type", "direct_appeal_status", "citation_role"],
        kind="stable",
    )
)

with pd.ExcelWriter(OUTPUT_XLSX, engine="openpyxl") as writer:
    final.to_excel(writer, sheet_name="classifications", index=False)
    summary.to_excel(writer, sheet_name="summary", index=False)

    for sheet_name in ("classifications", "summary"):
        ws = writer.sheets[sheet_name]
        ws.freeze_panes = "A2"
        ws.auto_filter.ref = ws.dimensions

print("Saved:", OUTPUT_XLSX)
print()
print("Classification status:")
display(
    final["classification_status"]
    .value_counts(dropna=False)
    .rename_axis("status")
    .reset_index(name="rows")
)
print()
display(summary)
```

## 7. Interpretation

The model works at the Court-document level for efficiency, while the final output
remains at the original candidate level: one row per normalized Commission identifier
in one Court document.

Each prompt includes the document opening (`BEGINNING_CHARS`), the closing
dispositif (`ENDING_CHARS`), and a ±`HIT_CONTEXT_CHARS` context window around
the first regex hit for every pending identifier. Additional hits are added greedily
up to the `MAX_PREPARED_DOCUMENT_CHARS` hard cap.

If the mandatory content (anchors + first-hit windows for every target) would already
exceed `MAX_PREPARED_DOCUMENT_CHARS`, the document is written to the checkpoint with
`classification_status = "deferred"` and skipped on subsequent reruns. This keeps
prompts within budget and avoids sending incomplete context windows.

The JSONL checkpoint is keyed by `job_id` (sha1 of
`record_id|reference_type|reference_normalized`). Interrupted runs resume without
repeating identifiers that were already classified successfully or deferred.
